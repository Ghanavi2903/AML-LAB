```python

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report

# Load dataset
df = pd.read_csv(r"C:\lab\AML-LAB\data\Social_Network_Ads.csv")

# Check dataset for missing values
print(df.isnull().sum())

# Convert Gender into numeric values
df["Gender"] = df["Gender"].map({
    "Female": 0,
    "Male": 1
})

# Features and target
X = df[["Gender", "Age", "EstimatedSalary"]]
y = df["Purchased"]

# Split data into training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

# Standardization
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)

# Check StandardScaler
print("Mean:", X_train.mean(axis=0))
print("Std:", X_train.std(axis=0))

# Train the model
classifier = LogisticRegression(random_state=0)
classifier.fit(X_train, y_train)

# Predict test data
y_pred = classifier.predict(X_test)

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)

# Accuracy
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy * 100, "%")

# Classification Report
print("Classification Report:")
print(classification_report(y_test, y_pred))

# Predict for a new person
new_person = np.array([[0, 30, 50000]])

# Scale using the existing scaler
new_person_scaled = sc.transform(new_person)

# Predict
prediction = classifier.predict(new_person_scaled)
print("Predicted Purchase (0 = No, 1 = Yes):", prediction[0])
```

```python

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report

# Load dataset
df = pd.read_csv(r"C:\lab\AML-LAB\data\Social_Network_Ads.csv")

# Check dataset for missing values
print(df.isnull().sum())

# Convert Gender into numeric values
df["Gender"] = df["Gender"].map({
    "Female": 0,
    "Male": 1
})

# Features and target
X = df[["Gender", "Age", "EstimatedSalary"]]
y = df["Purchased"]

# Split data into training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

# Standardization
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)

# Check StandardScaler
print("Mean:", X_train.mean(axis=0))
print("Std:", X_train.std(axis=0))

# Train the model
classifier=LogisticRegression(penalty="l1",solver="liblinear",C=1.0)
classifier.fit(X_train, y_train)
# Apply Sigmoid Function

# Calculate the linear score
z = classifier.decision_function(X_test)

# Sigmoid function
sigmoid = 1 / (1 + np.exp(-z))

# Display probabilities for the first 10 samples
print("Sigmoid probabilities:")
print(sigmoid[:10])

# Convert probabilities into classes using threshold 0.5
y_sigmoid_pred = (sigmoid >= 0.5).astype(int)

print("Predicted classes:")
print(y_sigmoid_pred[:10])
# Predict test data
y_pred = classifier.predict(X_test)

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)

# Accuracy
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy * 100, "%")

# Classification Report
print("Classification Report:")
print(classification_report(y_test, y_pred))

# Predict for a new person
new_person = np.array([[0, 30, 50000]])

# Scale using the existing scaler
new_person_scaled = sc.transform(new_person)

# Predict
prediction = classifier.predict(new_person_scaled)
print("Predicted Purchase (0 = No, 1 = Yes):", prediction[0])
```
