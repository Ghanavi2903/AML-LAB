```python
salary = np.array([25000, 32000, 28000, 45000,
                   50000, 38000, 42000, 30000])

print("Total:", np.sum(salary))
print("Average:", np.mean(salary))
print("Highest:", np.max(salary))
print("Lowest:", np.min(salary))
print("Median:", np.median(salary))
print("Standard deviation:", np.std(salary))
```

```python
marks = np.array([45, 78, 89, 32, 67, 90, 55, 76, 41, 84])

print("Marks greater than 75:", marks[marks > 75])

print("Marks between 50 and 80:",
      marks[(marks >= 50) & (marks <= 80)])

below_50 = np.sum(marks < 50)
print("Students below 50:", below_50)

above_60 = np.sum(marks >= 60)
percentage = (above_60 / len(marks)) * 100

print("Percentage scoring 60 or above:", percentage, "%")
```

```python
marks = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [88, 92, 85],
    [55, 60, 58]
])

total = np.sum(marks, axis=1)
average = np.mean(marks, axis=1)
subject_average = np.mean(marks, axis=0)
highest_student = np.argmax(total)
highest_subject = np.max(marks, axis=0)

print("Total marks:", total)
print("Average marks:", average)
print("Average in each subject:", subject_average)
print("Student with highest total:", highest_student + 1)
print("Highest mark in each subject:", highest_subject)
```

```python
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("Matrix addition:")
print(A + B)

print("\nElement-wise multiplication:")
print(A * B)

print("\nMatrix multiplication:")
print(A @ B)

print("\nTranspose of A:")
print(A.T)
```

A * B  → element-by-element multiplication

A @ B  → matrix multiplication

```python
temperature = np.array([
    31, 32, 33, 35, 36, 34, 32,
    30, 29, 31, 33, 35, 37, 36
])

average = np.mean(temperature)

print("Mean:", average)
print("Maximum:", np.max(temperature))
print("Minimum:", np.min(temperature))

print("Days above average:",
      np.sum(temperature > average))

print("Temperatures greater than 34:",
      temperature[temperature > 34])

print("Difference:",
      np.max(temperature) - np.min(temperature))

print("Standard deviation:",
      np.std(temperature))
```

```python
sales = np.array([
    12000, 15000, 17000, 14000,
    18000, 21000, 19000, 22000,
    25000, 23000, 26000, 28000
])

print("Total annual sales:", np.sum(sales))
print("Average monthly sales:", np.mean(sales))

highest_month = np.argmax(sales) + 1
lowest_month = np.argmin(sales) + 1

print("Month with highest sales:", highest_month)
print("Month with lowest sales:", lowest_month)

print("Months above 20,000:",
      np.sum(sales > 20000))

percentage_increase = ((sales[-1] - sales[0]) / sales[0]) * 100

print("Percentage increase:", percentage_increase, "%")
```

```python
marks = np.array([
    [78, 85, 90, 88],
    [65, 70, 72, 68],
    [92, 88, 95, 91],
    [55, 60, 58, 62],
    [80, 75, 82, 79]
])

total = np.sum(marks, axis=1)
average = np.mean(marks, axis=1)
subject_average = np.mean(marks, axis=0)

print("Total marks:", total)
print("Average of every student:", average)
print("Average of every subject:", subject_average)

highest_student = np.argmax(average) + 1
lowest_student = np.argmin(average) + 1

print("Highest-scoring student:", highest_student)
print("Lowest-scoring student:", lowest_student)

print("Students with average greater than 80:",
      np.where(average > 80)[0] + 1)

print("Overall class average:", np.mean(marks))

# Performance categories
category = np.select(
    [
        average >= 85,
        average >= 70,
        average >= 50
    ],
    [
        "Excellent",
        "Good",
        "Average"
    ],
    default="Needs Improvement"
)

print("Performance categories:", category)
```

```python
a = np.array([10, 20, 30, 40, 50])

normalized = (a - np.min(a)) / (np.max(a) - np.min(a))

print("Original:", a)
print("Normalized:", normalized)
```

```python
a = np.array([10, 20, 30, 40, 50])

standardized = (a - np.mean(a)) / np.std(a)

print("Original:", a)
print("Standardized:", standardized)
```

```python
import time
import numpy as np

# Python list
list1 = list(range(1000000))
list2 = list(range(1000000))

start = time.time()

result = [a * b for a, b in zip(list1, list2)]

list_time = time.time() - start


# NumPy array
array1 = np.arange(1000000)
array2 = np.arange(1000000)

start = time.time()

result = array1 * array2

numpy_time = time.time() - start


print("Python list time:", list_time)
print("NumPy array time:", numpy_time)
```
