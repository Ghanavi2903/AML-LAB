```python
print(df[df["Marks"]>80])
print(df[(df["Marks"]>=75) & (df["Marks"]<=90)])
print(df[df["Department"] == "CSE"])
print(df[(df["Department"] == "CSE") & (df["Marks"] > 80)])
```

```python
df["Result"] = df["Marks"].apply(lambda x: "Pass" if x >= 75 else "Fail")

print(df)
```

```python
df["Grade"] = df["Marks"].apply(
    lambda x: "A" if x >= 90
    else "B" if x >= 80
    else "C" if x >= 70
    else "D"
)

print(df)
```

```python
print("Marks in ascending order:")
print(df.sort_values("Marks"))

print("\nMarks in descending order:")
print(df.sort_values("Marks", ascending=False))

print("\nBy Department and then Marks:")
print(df.sort_values(["Department", "Marks"]))
```

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Asha", "Ravi", "Meena", "Kiran", "Arun"],
    "Age": [22, 23, np.nan, 24, 22],
    "Marks": [85, np.nan, 92, 74, 88]
}

students = pd.DataFrame(data)

print("Missing values:")
print(students.isnull())

print("\nNumber of missing values in each column:")
print(students.isnull().sum())

print("\nRows containing missing values:")
print(students[students.isnull().any(axis=1)])
students["Age"] = students["Age"].fillna(students["Age"].mean())

students["Marks"] = students["Marks"].fillna(students["Marks"].mean())

print(students)
```

```python
import pandas as pd

marks = pd.DataFrame({
    "Student": ["A", "B", "C", "D", "E"],
    "Marks": [78, 85, 92, 67, 88]
})

print("Mean:", marks["Marks"].mean())
print("Median:", marks["Marks"].median())
print("Maximum:", marks["Marks"].max())
print("Minimum:", marks["Marks"].min())
print("Standard deviation:", marks["Marks"].std())
print("Sum:", marks["Marks"].sum())
```

```python
data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE"
    ],

    "Marks": [
        85, 78, 92,
        74, 88, 80
    ]
}

df = pd.DataFrame(data)

print(df.groupby("Department")["Marks"].mean())
print(df.groupby("Department")["Marks"].max())
print(df.groupby("Department")["Name"].count())
print("\nMean, Minimum, Maximum and Count:")
print(df.groupby("Department")["Marks"].agg(
    ["mean", "min", "max", "count"]))
```

```python
data = {
    "Name": [
        "Asha", "Ravi", "Meena",
        "Kiran", "Arun", "Divya",
        "John", "Priya"
    ],

    "Department": [
        "CSE", "ECE", "CSE",
        "ME", "ECE", "CSE",
        "ME", "ECE"
    ],

    "Attendance": [
        92, 75, 96, 68,
        82, 88, 72, 95
    ],

    "Internal": [
        85, 72, 94, 65,
        80, 86, 70, 91
    ],

    "FinalMarks": [
        88, 74, 96, 62,
        83, 89, 68, 93
    ]
}

df = pd.DataFrame(data)
print(df.head(5))
print(df.shape)
print(df["FinalMarks"].mean())
print(df["FinalMarks"].max())
print(df["FinalMarks"].min())
print(df[df["FinalMarks"]>80])
print(df.groupby("Department")["FinalMarks"].mean())
```

```python
print(df[df["Attendance"]>80])
print(df.sort_values("FinalMarks",ascending=False))
df["Result"]=df["FinalMarks"].apply(lambda x:"Pass"if x>=70 else "Fail")
print(df)
```

```python
df["performance"]=df["FinalMarks"].apply(lambda x:"Excellent"if x>=90 else "Very good" if x>=80 else "Good" if x>=70 else "needs improvement")
print(df)
avg_marks = df.groupby("Department")["FinalMarks"].mean()

print(avg_marks)
print("Department with highest average:", avg_marks.idxmax())
```

```python
em = pd.DataFrame({
    "Employee": [
        "E01", "E02", "E03", "E04",
        "E05", "E06", "E07", "E08"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance",
        "HR", "Finance", "IT", "Finance"
    ],

    "Experience": [
        5, 3, 7, 4,
        6, 8, 2, 5
    ],

    "Salary": [
        60000, 45000, 75000, 65000,
        55000, 85000, 48000, 70000
    ]
})


print(em["Salary"].mean())
print(em["Salary"].min())
print(em["Salary"].max())
print(em[em["Salary"] == em["Salary"].max()])
print(em[em["Salary"]>60000])
```

```python
# Average salary for each department
print(em.groupby("Department")["Salary"].mean())

# Average experience for each department
print(em.groupby("Department")["Experience"].mean())
print(em.groupby("Department")["Employee"].count())
print(em.sort_values("Salary", ascending=False))
```

```python
print(em[(em["Department"] == "IT") & (em["Experience"] > 3)])
```

```python
em["SalaryLevel"] = em["Salary"].apply(
    lambda x: "High" if x >= 75000
    else "Medium" if x >= 50000
    else "Low"
)

print(em)
```
